```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model.train(
    data="/content/subset10k/content/UADETRAC/DETRAC_Upload/data/subsets/subset10k/dataset.yaml",
    epochs=25,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    project="/content/drive/MyDrive/new_UA/yolo_results10K",
    name="uadetrac_subset10k_e25",
    exist_ok=True
)
```

```python
model = YOLO("/content/drive/MyDrive/new_UA/yolo_results10K/uadetrac_subset10k_e50/weights/best.pt")

model.predict(
    source="/content/subset10k/content/UADETRAC/DETRAC_Upload/data/subsets/subset10k/images/val",
    imgsz=640,
    conf=0.25,
    project="/content/drive/MyDrive/new_UA/yolo_results10K",
    name="uadetrac_subset10k_e50_val",
)
```
