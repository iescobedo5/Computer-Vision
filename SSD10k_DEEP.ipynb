```python
import torch
import cv2
import time
import numpy as np
from pathlib import Path
from PIL import Image
from torchvision.transforms import functional as F
from deep_sort_realtime.deepsort_tracker import DeepSort
```

```python
IMAGE_DIR = Path("/content/drive/MyDrive/MVI_39031")

SAVE_DIR = Path("/content/drive/MyDrive/UA/ssd10k_deepsort/MVI_39031")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

TRACK_TXT = SAVE_DIR / "ssd10k_deepsort_pred.txt"

SSD_WEIGHTS = "/content/drive/MyDrive/UA/ssd_results/subset10k_25e/ssd_best_10k25e.pth"
```

```python
import torch
from torchvision.models.detection import ssd300_vgg16
from torchvision.models.detection.ssd import SSDClassificationHead

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

num_classes = 2

model = ssd300_vgg16(weights=None)

in_channels = [512, 1024, 512, 256, 256, 256]
num_anchors = model.anchor_generator.num_anchors_per_location()

model.head.classification_head = SSDClassificationHead(
    in_channels=in_channels,
    num_anchors=num_anchors,
    num_classes=num_classes
)

model.load_state_dict(torch.load(SSD_WEIGHTS, map_location=device))
model.to(device)
model.eval()

print("SSD model loaded.")
```

```python
tracker = DeepSort(
    max_age=30,
    n_init=3,
    max_cosine_distance=0.4,
    nn_budget=None
)
```

```python
CONF_THRESH = 0.25

image_files = sorted([
    p for p in IMAGE_DIR.iterdir()
    if p.suffix.lower() in [".jpg", ".jpeg", ".png"]
])

print(f"Found {len(image_files)} frames.")

start_time = time.time()

with open(TRACK_TXT, "w") as f:

    for frame_idx, img_path in enumerate(image_files, start=1):

        image = Image.open(img_path).convert("RGB")
        img_tensor = F.to_tensor(image).to(device)

        with torch.no_grad():
            output = model([img_tensor])[0]

        boxes = output["boxes"].detach().cpu().numpy()
        scores = output["scores"].detach().cpu().numpy()
        labels = output["labels"].detach().cpu().numpy()

        detections = []

        for box, score, label in zip(boxes, scores, labels):

            if score < CONF_THRESH:
                continue

            x1, y1, x2, y2 = box

            w = x2 - x1
            h = y2 - y1

            if w <= 0 or h <= 0:
                continue
            detections.append(([x1, y1, w, h], float(score), str(label)))

        tracks = tracker.update_tracks(detections, frame=np.array(image))

        for track in tracks:
            if not track.is_confirmed():
                continue

            track_id = track.track_id
            l, t, r, b = track.to_ltrb()

            w = r - l
            h = b - t

            f.write(
                f"{frame_idx},{track_id},{l:.2f},{t:.2f},{w:.2f},{h:.2f},1,-1,-1,-1\n"
            )

        if frame_idx % 100 == 0:
            print(f"Processed {frame_idx}/{len(image_files)} frames")

end_time = time.time()

total_time = end_time - start_time
fps = len(image_files) / total_time

print("Done!")
print(f"Saved tracking file to: {TRACK_TXT}")
print(f"Total time: {total_time:.2f} seconds")
print(f"FPS: {fps:.2f}")
```
